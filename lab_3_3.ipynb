```python
import pandas as pd
import spacy
url = "https://raw.githubusercontent.com/ramar92/NLP-Dataset/main/resume_dataset.csv"
df = pd.read_csv(url)
print("Dataset Loaded!")
df.head()
```

```python
sample_text = df['Resume_Text'].iloc[0]
print("/nSample Resume:\n", sample_text[:500], "...")
```

```python
nlp = spacy.load("en_core_web_sm")
# Process the text
doc = nlp(sample_text)
```

```python
print("\n---Token |POS| Depencdecy---")
for token in doc[:30]:
  print(f"{token.text:15} {token.pos_:10} {token.dep_}")
```

```python
print("\n-- Named Entities --")
for ent in doc.ents:
   print(ent.text,"->", ent.label_)
```

```python
print("\n-- Dependency Parse Struture--")
for token in doc:
  print(f"{token.text:15} <--{token.dep_}-- {token.head.text}")
```

```python
skills=[]
job_titles =[]
for ent in doc.ents:
  if ent.label_ in ["ORG", "PERSON", "WORK_OF_ART"]:
    job_titles.append(ent.text)
  if ent.label_ in ["SKILL", "PRODUCT", "NORP"]:
    skills.append(ent.text)

print("\nLikely Job Titles Found:", job_titles)
print("Likely Skills Found:", skills)
```
