```python
# Import necessary library
import numpy as np
```

```python
# Define two vectors
vector_a = np.array ([5,6,7])
vector_b = np.array ([8,9,10])

# perform vector addition
vector_addition = vector_a + vector_b
print("Vector Addition:", vector_addition)

# perform vector subraction
vector_subraction = vector_a - vector_b
print("Vector Subraction:", vector_subraction)

# Compute dot product
dot_product = np.dot(vector_a, vector_b)
print("Dot Product:", dot_product)
```
