```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv('test_Y3wMUE5_7gLdaTN.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.drop('Loan_ID' , axis=1 , inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.describe()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x='Credit_History' , data=df)
plt.title('Loan_Status Distribution')
plt.show()
```

```python
plt.figure(figsize=(10,3))
sns.countplot(x='Property_Area', hue='Credit_History' , data=df)
plt.title('Property Area vs Loan Status')
plt.show()
```

```python
plt.figure(figsize=(10,3))
sns.countplot(x='Education', hue='Credit_History' , data=df)
plt.title('Education vs Loan Status')
plt.show()
```

```python
plt.figure(figsize=(10,3))
sns.boxplot(x="Education" , y="ApplicantIncome" , data=df)
plt.title('Applicant_Income by Education')
plt.show()
```

```python
plt.figure(figsize=(10,3))
sns.boxplot(x="ApplicantIncome" , y="LoanAmount" , data=df)
plt.title('Loan Amount by ApplicantIncome')
plt.show()
```

```python
corr=df.corr(numeric_only=True)
plt.figure(figsize=(10,5))
sns.heatmap(corr , annot=True , cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
numerical_cols = ['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount', 'Loan_Amount_Term', 'Credit_History']
sns.pairplot(df[numerical_cols + ['Education']], hue='Education')
plt.show()
```

```python
plt.figure(figsize=(10,3))
sns.scatterplot(x='ApplicantIncome' , y='LoanAmount', data=df)
plt.title('Income vs Loan Amount')
plt.show()
```
